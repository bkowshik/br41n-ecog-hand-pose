# br41n-ecog-hand-pose

> Analyze an ECoG BCI dataset from an epilepsy patient performing rock-paper-scissors hand gestures, and optimize pre-processing, feature extraction, and classification against state-of-the-art baselines.

## Dataset

`data/ECoG_Handpose.mat` (provided by the BR41N.IO hackathon organizers):

- **Paradigm:** 90 trials of rock / paper / scissors gestures, cued for 2s and followed by a 2–3s rest screen.
- **Sampling rate:** 1200 Hz.
- **Recording site:** Right sensorimotor cortex, 60-channel ECoG grid spanning pre-/post-central, supra-marginal, inferior parietal, and middle/superior frontal gyri.
- **Channel layout** (`[channel × time]`):
  - CH1 — sample time
  - CH2–61 — ECoG (raw, DC-coupled), 60 channels
  - CH62 — paradigm label (`0`=relax, `1`=fist, `2`=peace, `3`=open hand)
  - CH63–67 — data glove (thumb, index, middle, ring, little finger flexion)

## Notebook pipeline

Each notebook in `nbs/` focuses on one step of the analysis. Functions defined in earlier notebooks are exported via [nbdev](https://nbdev.fast.ai/) into the `br41n_ecog_hand_pose` package and reused in later notebooks.

| Notebook | Module | Purpose |
|---|---|---|
| `01_data_loading` | `data` | Load `.mat`, expose ECoG / labels / glove accessors |
| `02_eda` | — | Visualize raw signals, label timeline, channel stats |
| `03_preprocessing` | `preprocessing` | Notch, bandpass, common average reference |
| `04_epoching` | `epoching` | Cue-locked trial segmentation from CH62 |
| `05_features` | `features` | Band power, time-frequency, CSP |
| `06_classification` | `classification` | LDA / SVM / Riemannian baselines |
| `07_deep_learning` | `deep` | EEGNet / ShallowConvNet end-to-end models |
| `08_finger_regression` | `regression` | Continuous decoding of glove channels |
| `09_benchmark` | — | Compare results against published SOTA |

## Developer guide

One-time bootstrap:

```sh
# 1. export notebooks → .py modules so the package is importable
$ nbdev-export

# 2. install the package in editable mode
$ pip install -e .

# 3. preview the rendered docs locally
$ nbdev-preview
```

Day-to-day, after editing any notebook in `nbs/`, sync code + docs + tests with:

```sh
$ nbdev-prepare
```
