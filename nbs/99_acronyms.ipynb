# Acronyms

> Glossary of the acronyms used across this project's notebooks, README, and module docstrings.

| Acronym | Expansion | Notes |
|---|---|---|
| BCI | Brain–Computer Interface | The class of system this dataset is built for. |
| ECoG | Electrocorticography | Cortical-surface electrodes; higher SNR / bandwidth than scalp EEG. |
| EEG | Electroencephalography | Scalp electrodes; relevant because most baseline models (EEGNet etc.) come from EEG literature. |
| CH | Channel | As in CH1–CH67 of the data matrix. |
| DC | Direct Current | The recording is DC-coupled, so very low frequencies are preserved (and need explicit removal). |
| Hz | Hertz | Cycles per second (sampling rate, frequency bands). |
| FS | Sampling Frequency | 1200 Hz here. |
| CAR | Common Average Reference | Re-reference each channel by subtracting the across-channel mean. |
| PSD | Power Spectral Density | Power per frequency bin. |
| STFT | Short-Time Fourier Transform | Sliding-window FFT used for time-frequency features. |
| CSP | Common Spatial Patterns | Spatial filter that maximizes variance ratio between classes. |
| ICA | Independent Component Analysis | Optional artifact-removal step. |
| ML | Machine Learning | Classical models — LDA, SVM, logistic regression, Riemannian. |
| DL | Deep Learning | End-to-end neural-network models on raw epochs. |
| LDA | Linear Discriminant Analysis | Classic linear classifier; common BCI baseline. |
| SVM | Support Vector Machine | Margin-based classifier. |
| CNN | Convolutional Neural Network | Backbone of EEGNet / ShallowConvNet. |
| EEGNet | EEG-specific compact CNN | Standard deep baseline for EEG/ECoG decoding. |
| MSE | Mean Squared Error | Regression loss / metric (used in `08_finger_regression`). |
| CV | Cross-Validation | k-fold or leave-one-out evaluation protocol. |
| SOTA | State Of The Art | Published best results we benchmark against in `09_benchmark`. |
